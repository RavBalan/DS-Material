```python
s = [1,2,2,3,3,3,3,4,4]
count = {}

for i in s:
    if i not in count:
        count[i] = 1        
    else:
        count[i] += 1
print(count)
```

```python
##MAP
def check_odd(num):
    return int(num)%2 ==0

num = [1,2,3,4,5,6,7,8]

print(list(map(check_odd,num)))
print(list(filter(check_odd,num)))
```

```python
lst = ['Ravi', 11, 'karthi', 44]

dic = {lst[i]: lst[i+1] for i in range(0, len(lst),2)}

print(dic)
```
