```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([1, 2, 3, 4])  # Independent variable
y = np.array([50, 55, 60, 65])  # Dependent variable

# Parameters
m, c = 0, 0  # Initial guess for slope and intercept
learning_rate = 0.01
iterations = 2
n = len(x)

# Gradient Descent
cost_history = []  # To store the cost at each iteration
for _ in range(iterations):
    # Predicted values
    y_pred = m * x + c

    # Compute cost (Mean Squared Error)
    cost = (1 / n) * np.sum((y - y_pred) ** 2)
    print(cost)
    cost_history.append(cost)

    # Gradients
    dm = -(2 / n) * np.sum(x * (y - y_pred))
    dc = -(2 / n) * np.sum(y - y_pred)

    # Update parameters
    m = m - learning_rate * dm
    c = c - learning_rate * dc

print(cost_history[0])
# Results
print(f"Final Slope (m): {m}")
print(f"Final Intercept (c): {c}")
# Print gradients at the final iteration
print(f"Gradient dm: {dm}")
print(f"Gradient dc: {dc}")
```

```python
import numpy as np

# Sample data
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Hyperparameters
alpha = 0.1  # Learning rate
iterations = 1000  # Number of iterations

# Initial parameters (m and c)
m = 0
c = 0

# Number of data points
m_value = len(x)

# Hypothesis function (predict y values using current m and c)
def hypothesis(x, m, c):
    return m * x + c

# Cost function (Mean Squared Error)
def cost_function(x, y, m, c):
    predictions = hypothesis(x, m, c)
    errors = predictions - y
    cost = (1 / (2 * m_value)) * np.sum(errors ** 2)
    return cost

# Gradient Descent function
def gradient_descent(x, y, m, c, alpha, iterations):
    cost_history = []

    for _ in range(iterations):
        predictions = hypothesis(x, m, c)

        # Compute gradients
        dm = (1 / m_value) * np.sum((predictions - y) * x)
        dc = (1 / m_value) * np.sum(predictions - y)

        # Update parameters
        m = m - alpha * dm
        c = c - alpha * dc

        # Calculate cost and save it
        cost = cost_function(x, y, m, c)
        cost_history.append(cost)

    return m, c, cost_history

# Run gradient descent to optimize m and c
m, c, cost_history = gradient_descent(x, y, m, c, alpha, iterations)

print(f"Final Slope (m): {m}")
print(f"Final Intercept (c): {c}")
print(f"Final Cost: {cost_history[-1]}")

# Plotting the cost function over iterations
import matplotlib.pyplot as plt
plt.plot(range(iterations), cost_history, color='blue')
plt.xlabel('Iterations')
plt.ylabel('Cost')
plt.title('Cost Function Convergence')
plt.show()
```
