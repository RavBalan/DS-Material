```python
salaryData = pd.DataFrame({'Employee': np.random.choice(['Rajan','Sentihl','Guna','Anbhu'],size=(10)),'Salaray': np.random.choice([1000,2000,3000], size=(10))})
employeeData = pd.DataFrame({'EmployeeName': np.random.choice(['Rajan','Rajan','Sentihl','Guna','Anbhu'],size=(10)),'Salaray': np.random.choice(['salem','Chennai','Kochin'], size=(10))})
employeeData = employeeData.drop_duplicates(subset='EmployeeName')

Data = pd.merge(salaryData,employeeData, left_on='Employee',right_on='EmployeeName',how='inner')
Data
```
