Practice Fro Nominol Encoding

```python
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
df = sns.load_dataset('tips')
df[['tip']]#.count()

df[df['tip']==1.0]

# for group , name in df.groupby('tip'):
#     print(group)

```

```python
encoder = OneHotEncoder()
```

```python
encoded = encoder.fit_transform(df[['tip']]).toarray()
encoded = pd.DataFrame(encoded,columns=encoder.get_feature_names_out())

ouptut = pd.concat([df['tip'],encoded],axis=1)
ouptut
```

Ordinal Encoding:

```python
mean_total_bill = df.groupby('time')['total_bill'].mean().to_dict()

df['mean_total_bill'] = df['time'].map(mean_total_bill)
print(df[['time','total_bill','mean_total_bill']])
```
