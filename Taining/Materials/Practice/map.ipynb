```python
import folium
from IPython.display import display
import folium.map

dot1 = [12.937585046406076, 80.14490239031306]
mymap = folium.Map(location=dot1,zoom_control=True)

folium.Marker(
    dot1,
    popup="New York",
    icon=folium.Icon(color="orange",icon="info-sign")).add_to(mymap)
display(mymap)
```
